```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv("E:\\Coding\\Portfolio Projects\\Customer Journey Analysis, Intervention and Performance Tracking\\dropoff_project\\data\\user_journey_raw.csv")
df
```

```python
df.drop(columns=['converted', 'exit_stage', 'reached_stage2', 'reached_stage3', 'reached_stage4', 'reached_stage5'], inplace=True)
df
```

```python
#Converted Customers

df['converted'] = df['conversion_date'].notna().astype(int)
df
```

```python
#Exit Stage

import numpy as np
exit = [
    df['stage2_date'].isna(),
    df['stage3_date'].isna(),
    df['stage4_date'].isna(),
    df['stage5_date'].isna(),
    df['conversion_date'].isna()
]
choices = [
    'stage1',
    'stage2',
    'stage3',
    'stage4',
    'stage5'
]
df['exit_stage'] = np.select(exit, choices, default='Converted')
df
```

```python
#Stage Progression
df['stage2_reached'] = df['stage2_date'].notna().astype(int)
df['stage3_reached'] = df['stage3_date'].notna().astype(int)
df['stage4_reached'] = df['stage4_date'].notna().astype(int)
df['stage5_reached'] = df['stage5_date'].notna().astype(int)
df
```

```python
df_pre = df[df['period']=='Pre-Intervention'].copy()
df_pre = df_pre.drop(columns='period')
df_pre
```

```python
df_post = df[df['period']=='Post-Intervention'].copy()
df_post = df_post.drop(columns='period')
df_post
```

```python
df_pre.to_csv('first_transformations.csv', index=False)
```

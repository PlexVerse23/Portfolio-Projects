```python
import pandas as pd
import numpy as np
from scipy import stats
from scipy.stats import chi2_contingency, norm
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("E:\\Coding\\Portfolio Projects\\Customer Journey Analysis, Intervention and Performance Tracking\\dropoff_project\\data\\user_journey_raw.csv")
df
```

```python
pre = df[df['period']=='Pre-Intervention']
post = df[df['period']=='Post-Intervention']

pre_conv = pre['converted'].sum()
pre_cvr = pre_conv/len(pre)

post_conv = post['converted'].sum()
post_cvr = post_conv/len(post)

print(f"Pre CVR : {pre_cvr:.2%}")
print(f"Post CVR : {post_cvr:.2%}")
```

```python
#Z-Test

pool = (pre_conv + post_conv) / (len(pre) + len(post))

z = (post_cvr - pre_cvr) / np.sqrt(pool * (1 - pool) * (1/len(pre) + 1/len(post)))

p_value = 2 * (1 - norm.cdf(abs(z)))

print(f"Z-score: {z:.3f}")
print(f"P-value: {p_value:.6f}")
```

```python
#Chi-Square Test

stage3_users = df[df['reached_stage3'] == 1]

contingency = pd.crosstab(stage3_users['segment'], 
                           stage3_users['reached_stage4'])

chi2, p_value, dof, expected = chi2_contingency(contingency)

print(f"Chi-square: {chi2:.3f}")
print(f"P-value: {p_value:.6f}")
print(f"Degrees of freedom: {dof}")
```
